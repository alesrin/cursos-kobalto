axis=0 suma columnas (resultado: 1 valor por columna)

np.argmax(array) devuelve el INDICE del valor maximo, no el valor.

## Ejercicio 2: Aplicacion de Descuentos

**Objetivo:** Broadcasting con vectores fila

La tienda aplica descuentos diferentes cada mes (rebajas, Black Friday, etc.). Debes calcular el impacto en los ingresos.



### Tareas

- Calcula los precios con descuento para cada mes: precio * (1 - descuento)
- Calcula los ingresos CON descuento aplicado
- Calcula cuanto dinero se 'pierde' por los descuentos
- Que mes tiene la mayor perdida por descuentos?
- Que porcentaje de los ingresos totales se pierde en descuentos?


````python
# EJERCICIO 2: Aplicacion de descuentos
# Usa precios_col e ingresos del ejercicio anterior

# 1. Precios con descuento para cada mes
# Pista: factor_descuento = 1 - descuentos


# 2. Ingresos con descuento aplicado


# 3. Perdida por descuentos (diferencia)


# 4. Mes con mayor perdida


# 5. Porcentaje perdido en descuentos

````

<details>
<summary><strong>💡 Pistas (click para expandir)</strong></summary>

**Pista 1: Broadcasting automatico**

descuentos tiene forma (12,). Cuando operas con precios_col (4,1), NumPy expande:
- descuentos: (12,) -> (1, 12) -> (4, 12)
- precios_col: (4, 1) -> (4, 12)
Resultado: (4, 12)

**Pista 2: Factor de descuento**

Si descuento = 0.20 (20%), el factor es:
1 - 0.20 = 0.80
Precio final = precio * 0.80

**Pista 3: Perdida por descuentos**

perdida = ingresos_sin_descuento - ingresos_con_descuento
Luego suma todo: perdida.sum()

**Pista 4: Porcentaje**

porcentaje = (perdida_total / ingresos_totales) * 100

</details>


<details>
<summary><strong>✅ Solución (click para ver)</strong></summary>

```python
# EJERCICIO 2: Aplicacion de descuentos

# 1. Precios con descuento para cada mes
# precios_col: (4, 1), descuentos: (12,) -> resultado: (4, 12)
factor_descuento = 1 - descuentos
precios_mensuales = precios_col * factor_descuento
print(f"Forma precios_mensuales: {precios_mensuales.shape}")

# Mostrar precios del Portatil cada mes
print(f"\nPrecios Portatil por mes:")
print(precios_mensuales[0].round(2))

# 2. Ingresos con descuento
ingresos_descuento = unidades * precios_mensuales
print(f"\nForma ingresos_descuento: {ingresos_descuento.shape}")

# 3. Perdida por descuentos
perdida = ingresos - ingresos_descuento
print(f"\nPerdida total por descuentos: {perdida.sum():.2f} EUR")

# 4. Mes con mayor perdida
perdida_por_mes = perdida.sum(axis=0)
mes_max_perdida = np.argmax(perdida_por_mes)
print(f"Mes con mayor perdida: {meses[mes_max_perdida]} ({perdida_por_mes[mes_max_perdida]:.2f} EUR)")

# 5. Porcentaje perdido
porcentaje_perdido = (perdida.sum() / ingresos.sum()) * 100
print(f"Porcentaje perdido en descuentos: {porcentaje_perdido:.2f}%")
```

## Ejercicio 3: Analisis de Beneficios

**Objetivo:** Combinar múltiples operaciones de broadcasting

Calcula el beneficio real considerando precios, descuentos y costes. Identifica situaciones problematicas.



### Tareas

- Calcula el margen unitario por producto: precio - coste
- Calcula el margen unitario CON descuento para cada mes
- Calcula el beneficio total (margen * unidades) para cada producto/mes
- Identifica los meses donde algun producto tiene margen negativo
- Calcula el beneficio total anual por producto
- EXTRA: Normaliza los beneficios mensuales de cada producto


````python
# EJERCICIO 3: Analisis de beneficios
# Usa las variables de ejercicios anteriores

# 1. Margen unitario por producto (sin descuento)


# 2. Margen con descuento para cada mes
# Pista: necesitas costes_col


# 3. Beneficio total por producto/mes


# 4. Hay margenes negativos? Donde?
# Pista: usa np.where()


# 5. Beneficio anual por producto


# 6. EXTRA: Normalizacion con broadcasting
# Pista: usa keepdims=True

````

<details>
<summary><strong>💡 Pistas (click para expandir)</strong></summary>

**Pista 1: Margen con descuento**

margen_descuento = precios_mensuales - costes_col

Donde:
- precios_mensuales: (4, 12) - ya tiene descuentos
- costes_col: costes.reshape(4, 1) -> (4, 1) -> (4, 12)

**Pista 2: np.where para encontrar posiciones**

np.where(condición) devuelve tupla de arrays con índices.

Ejemplo:
filas, cols = np.where(matriz < 0)
for f, c in zip(filas, cols):
    print(f'Fila {f}, Col {c}')

**Pista 3: any() para verificar condición**

array.any() -> True si ALGUN elemento es True
array.all() -> True si TODOS son True

(margen < 0).any() -> Hay algun margen negativo?

**Pista 4: Normalizacion**

z = (x - media) / std

Con keepdims=True mantienes la forma para broadcasting:
media = arr.mean(axis=1, keepdims=True)  # (4, 1)
std = arr.std(axis=1, keepdims=True)     # (4, 1)

**Pista 5: keepdims explicado**

Sin keepdims:
arr.mean(axis=1) -> forma (4,)

Con keepdims=True:
arr.mean(axis=1, keepdims=True) -> forma (4, 1)

La forma (4, 1) permite broadcasting con (4, 12)

</details>


<details>
<summary><strong>✅ Solución (click para ver)</strong></summary>

```python
# EJERCICIO 3: Analisis de beneficios

# 1. Margen unitario por producto (sin descuento)
margen_base = precios - costes
print(f"Margen base por producto: {margen_base}")

# 2. Margen con descuento para cada mes
costes_col = costes.reshape(4, 1)
margen_descuento = precios_mensuales - costes_col
print(f"\nForma margen_descuento: {margen_descuento.shape}")

# 3. Beneficio total por producto/mes
beneficio = margen_descuento * unidades
print(f"Forma beneficio: {beneficio.shape}")

# 4. Meses con margen negativo
margen_negativo = margen_descuento < 0
print(f"\nHay margenes negativos? {margen_negativo.any()}")

if margen_negativo.any():
    filas, cols = np.where(margen_negativo)
    for f, c in zip(filas, cols):
        print(f"  ALERTA: {productos[f]} en {meses[c]}: margen = {margen_descuento[f, c]:.2f} EUR")
else:
    print("  Todos los margenes son positivos")

# 5. Beneficio anual por producto
beneficio_anual = beneficio.sum(axis=1)
print(f"\nBeneficio anual por producto:")
for i, prod in enumerate(productos):
    print(f"  {prod}: {beneficio_anual[i]:.2f} EUR")

# 6. EXTRA: Normalizacion con broadcasting
media_producto = beneficio.mean(axis=1, keepdims=True)
std_producto = beneficio.std(axis=1, keepdims=True)
beneficio_norm = (beneficio - media_producto) / std_producto

print(f"\nBeneficio normalizado (primeras 3 columnas):")
print(beneficio_norm[:, :3].round(2))
```
