import matplotlib.pyplot as plt

# EJERCICIO 4: Anotaciones y Marcadores

import matplotlib.pyplot as plt
import numpy as np

# Generar datos de precio de accion
np.random.seed(42)
dias = np.arange(1, 253)
precio_base = 150
cambios = np.random.randn(252) * 2
precio = precio_base + np.cumsum(cambios)

# Crear figura
fig, ax = plt.subplots(figsize=(14, 7))

# Dibujar línea de precio
ax.plot(dias, precio, color='#2c3e50', linewidth=1.5, label='Precio ACME Corp')

# Encontrar máximo y mínimo
dia_max = np.argmax(precio)
precio_max = precio[dia_max]
dia_min = np.argmin(precio)
precio_min = precio[dia_min]

# Precio medio
precio_medio = np.mean(precio)

# Anotación para el máximo
ax.annotate(f'Máximo: {precio_max:.2f} EUR',
            xy=(dia_max, precio_max),
            xytext=(dia_max + 30, precio_max + 3),
            arrowprops=dict(arrowstyle='->', color='#27ae60', lw=2),
            fontsize=11, fontweight='bold', color='#27ae60')

# Anotación para el mínimo
ax.annotate(f'Mínimo: {precio_min:.2f} EUR',
            xy=(dia_min, precio_min),
            xytext=(dia_min + 30, precio_min - 5),
            arrowprops=dict(arrowstyle='->', color='#e74c3c', lw=2),
            fontsize=11, fontweight='bold', color='#e74c3c')

# Línea horizontal con precio medio
ax.axhline(y=precio_medio, color='#7f8c8d', linestyle='--',
           alpha=0.8, linewidth=2, label=f'Media: {precio_medio:.2f} EUR')

# Zona sombreada de alta volatilidad
ax.axvspan(60, 90, alpha=0.15, color='#e74c3c', label='Alta volatilidad')

# Puntos destacados
ax.scatter([dia_max], [precio_max], color='#27ae60', s=100, zorder=5)
ax.scatter([dia_min], [precio_min], color='#e74c3c', s=100, zorder=5)

# Título, etiquetas y leyenda
ax.set_title('Evolución del Precio ACME Corp - 2024',
             fontsize=16, fontweight='bold')
ax.set_xlabel('Dia de Trading', fontsize=12)
ax.set_ylabel('Precio (EUR)', fontsize=12)
ax.legend(loc='upper left', fontsize=10)
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# EJERCICIO 5: Personalización de Estilo

import matplotlib.pyplot as plt
import numpy as np

# Datos de ventas mensuales
meses = ['Ene', 'Feb', 'Mar', 'Abr', 'May', 'Jun']
ventas_2023 = [45, 52, 48, 61, 55, 67]
ventas_2024 = [51, 58, 62, 70, 68, 78]

# Colores corporativos
color_primary = '#1a365d'
color_secondary = '#2b6cb0'
color_accent = '#48bb78'

# Crear figura
fig, ax = plt.subplots(figsize=(12, 6))

# Configurar colores de fondo
fig.patch.set_facecolor('#f7fafc')
ax.set_facecolor('#ffffff')

# Dibujar líneas con estilo corporativo

        color=color_secondary, label='2023', alpha=0.7)
ax.plot(meses, ventas_2024, marker='o', linewidth=2.5, markersize=8,
        color=color_primary, label='2024')

# Rellenar área entre líneas
ax.fill_between(meses, ventas_2023, ventas_2024,
                alpha=0.1, color=color_accent)

# Personalizar spines
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_color('#cbd5e0')
ax.spines['bottom'].set_color('#cbd5e0')

# Personalizar grid
ax.grid(True, axis='y', alpha=0.3, linestyle='-', color='#e2e8f0')
ax.set_axisbelow(True)

# Calcular estadísticas
crecimiento = [(v24 - v23) / v23 * 100 for v23, v24 in zip(ventas_2023, ventas_2024)]
crecimiento_medio = np.mean(crecimiento)

# Caja de texto con estadísticas

             edgecolor='#cbd5e0', alpha=0.9)

        verticalalignment='top', bbox=props, color=color_primary)

import matplotlib.pyplot as plt

# EJERCICIO 8: Informe Visual Completo

import matplotlib.pyplot as plt
import numpy as np
from matplotlib.gridspec import GridSpec

# Datos
np.random.seed(42)
meses = ['Ene', 'Feb', 'Mar', 'Abr', 'May', 'Jun', 'Jul', 'Ago', 'Sep', 'Oct', 'Nov', 'Dic']
ingresos = [120, 135, 142, 155, 148, 162, 175, 168, 182, 195, 205, 220]
productos = ['Software', 'Servicios', 'Hardware', 'Soporte']
ventas_producto = [45, 30, 15, 10]
regiones = ['Norte', 'Sur', 'Este', 'Oeste']
ventas_region = [350, 280, 220, 180]

# KPIs
total_ingresos = sum(ingresos) * 1000
crecimiento = ((ingresos[-1] - ingresos[0]) / ingresos[0]) * 100
mejor_mes = meses[np.argmax(ingresos)]

# Colores corporativos
COLORS = {
    'primary': '#1a365d',
    'secondary': '#3182ce',
    'accent': '#48bb78',
    'warning': '#ed8936',
    'bg': '#f7fafc'
}

# Crear figura con GridSpec
fig = plt.figure(figsize=(16, 12))
fig.patch.set_facecolor(COLORS['bg'])
gs = GridSpec(4, 3, figure=fig, height_ratios=[0.8, 1.5, 1.5, 2], hspace=0.3, wspace=0.3)

# Fila de KPIs
kpi_data = [
    (f"${total_ingresos/1000000:.1f}M", "Ingresos Totales", COLORS['primary']),
    (f"+{crecimiento:.1f}%", "Crecimiento Anual", COLORS['accent']),
    (mejor_mes, "Mejor Mes", COLORS['warning'])
]

for i, (valor, label, color) in enumerate(kpi_data):
    ax_kpi = fig.add_subplot(gs[0, i])
    ax_kpi.set_facecolor('white')
    ax_kpi.axis('off')
    ax_kpi.text(0.5, 0.65, valor, ha='center', va='center',
                fontsize=28, fontweight='bold', color=color)
    ax_kpi.text(0.5, 0.25, label, ha='center', va='center',
                fontsize=12, color='#718096')
    # Borde
    for spine in ax_kpi.spines.values():
        spine.set_visible(True)
        spine.set_color('#e2e8f0')

# Gráfico principal de ingresos (span 2 filas)
ax_main = fig.add_subplot(gs[1:3, :])
ax_main.set_facecolor('white')

ax_main.fill_between(meses, ingresos, alpha=0.3, color=COLORS['secondary'])
ax_main.plot(meses, ingresos, marker='o', linewidth=3, markersize=8,
             color=COLORS['primary'], markerfacecolor='white', markeredgewidth=2)

# Anotación del máximo
max_idx = np.argmax(ingresos)
ax_main.annotate(f'Máximo: {ingresos[max_idx]}K',
                 xy=(meses[max_idx], ingresos[max_idx]),
                 xytext=(meses[max_idx], ingresos[max_idx] + 15),
                 fontsize=10, fontweight='bold', color=COLORS['accent'],
                 ha='center')

ax_main.set_title('Evolución de Ingresos Mensuales', fontsize=14, fontweight='bold',
                  color=COLORS['primary'], pad=10)
ax_main.set_ylabel('Ingresos (miles EUR)', fontsize=11)
ax_main.grid(True, alpha=0.3)
ax_main.spines['top'].set_visible(False)
ax_main.spines['right'].set_visible(False)

# Donut de productos
ax_donut = fig.add_subplot(gs[3, 0])
colors_pie = ['#3182ce', '#48bb78', '#ed8936', '#9f7aea']
wedges, texts, autotexts = ax_donut.pie(ventas_producto, labels=productos,
                                         autopct='%1.0f%%', colors=colors_pie,
                                         wedgeprops=dict(width=0.5, edgecolor='white'),
                                         pctdistance=0.75)
for autotext in autotexts:
    autotext.set_fontsize(9)
    autotext.set_fontweight('bold')
    autotext.set_color('white')

ax_donut.set_title('Ventas por Producto', fontsize=12, fontweight='bold',
                   color=COLORS['primary'])

# Barras de regiones
ax_bar = fig.add_subplot(gs[3, 1:])
ax_bar.set_facecolor('white')
bars = ax_bar.barh(regiones, ventas_region, color=colors_pie, edgecolor='white', height=0.6)

# Valores en las barras
for bar, val in zip(bars, ventas_region):
    ax_bar.text(val + 5, bar.get_y() + bar.get_height()/2, f'{val}K',
                va='center', fontsize=10, fontweight='bold', color=COLORS['primary'])

ax_bar.set_title('Ventas por Región', fontsize=12, fontweight='bold',
                 color=COLORS['primary'])
ax_bar.set_xlabel('Ventas (miles EUR)', fontsize=10)
ax_bar.spines['top'].set_visible(False)
ax_bar.spines['right'].set_visible(False)
ax_bar.set_xlim(0, max(ventas_region) * 1.15)

# Título general
fig.suptitle('INFORME EJECUTIVO - Q4 2024',
             fontsize=20, fontweight='bold', color=COLORS['primary'], y=0.98)

plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
